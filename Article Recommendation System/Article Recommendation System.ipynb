```python
# Importing Libraries
import pandas as pd
import numpy as np
from sklearn.feature_extraction import text
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Loading the Dataset
data = pd.read_csv('Article Recommendation System.csv', encoding='latin-1') 
```

```python
# Analyse the Top 5 rows of the data
data.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Assume `data` already contains your DataFrame with an 'Article' and 'Title' column
articles = data["Article"].astype(str).tolist()

# Create the TF-IDF matrix
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(articles)

# Compute cosine similarity matrix
similarity_matrix = cosine_similarity(tfidf_matrix)

# Function to get top 4 similar article indices (excluding itself)
def recommend_articles(similarity_row, index):
    similar_indices = similarity_row.argsort()[-6:-1][::-1]  # Top 5 excluding itself
    similar_titles = data["Title"].iloc[similar_indices]
    return ", ".join(similar_titles)

# Generate recommendations
data["Recommended Articles"] = [
    recommend_articles(sim_row, idx) for idx, sim_row in enumerate(similarity_matrix)
]

# View the result
data.head()
```
