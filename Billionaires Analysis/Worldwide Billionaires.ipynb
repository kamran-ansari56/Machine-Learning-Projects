```python
# Importing the Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Loading the Data
data = pd.read_csv('Billionaires Analysis.csv')
```

```python
# Analyse the Top 5 rows of the data
data.head()
```

```python
# Now let's check for any null values
data.isnull().sum()
```

```python
# So this dataset contains 79 missing values, let's remove these rows
data.dropna(inplace= True)
```

##### The NetWorth column in this dataset has a $ sign at the beginning of Billionaires’ Net worth and B at the end.

```python
# So let's remove these signs and convert the NetWorth column to float
data['NetWorth'] = data['NetWorth'].str.strip('$')
data['NetWorth'] = data['NetWorth'].str.strip('B')
data['NetWorth'] = data['NetWorth'].astype(float)
```

```python
# Now, look at the top 10 billionaires' Net Worth
df = data.sort_values(by = ['NetWorth'], ascending= False).head(10)
```

```python
plt.figure(figsize = (20,10))
sns.histplot(x = df['Name'], hue = df['NetWorth'], data = df)
plt.show()
```

```python
# Now let’s have a look at the top 5 domains with the most number of billionaires
a= data['Source'].value_counts().head()
index = a.index
sources = a.values
```

```python
custom_colors = ["skyblue", "yellowgreen", 'tomato', "blue", "red"]
plt.figure(figsize=(5, 5))
plt.pie(sources, labels=index, colors=custom_colors)
central_circle = plt.Circle((0, 0), 0.5, color='white')
fig = plt.gcf()
fig.gca().add_artist(central_circle)
plt.rc('font', size=12)
plt.title("Top 5 Domains to Become a Billionaire", fontsize=20)
plt.show()
```
