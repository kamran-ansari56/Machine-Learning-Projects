#### Importing the Libraries


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

#### Load the Dataset

```python
df = pd.read_csv('Breast Cancer Wisconsin Diagnosis.xls')
```

#### Anlayse the Top 5 rows of the Data

```python
df.head()
```

```python
# Now let's check for more information about the data
df.info()
```

```python
# We will drop columns - 'id' and 'Unnamed: 32' as they have no role in prediction
df.drop(columns= ['Unnamed: 32', 'id'], inplace= True)
```

```python
df
```

```python
# Now let's Convert the diagnosis value of M and B to a numerical value where M (Malignant) = 1 and B (Benign) = 0
def diagnosis_values(diagnosis):
    if diagnosis == 'M':
        return 1
    else:
        return 0

df['diagnosis'] = df['diagnosis'].apply(diagnosis_values)
```

```python
sns.lmplot(x = 'radius_mean', y = 'texture_mean', hue = 'diagnosis', data = df)
plt.show
```

```python
sns.lmplot(x ='smoothness_mean', y = 'compactness_mean', 
           data = df, hue = 'diagnosis')
```

```python
# Input and Output data
X = np.array(df.iloc[:, 1:])
y = np.array(df['diagnosis'])
```

```python
# Splitting data to training and testing
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.33, random_state= 42)
```

```python
# Using KNN
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors= 13)
knn.fit(X_train, y_train)
```

```python
knn.score(X_test, y_test)
```

#### Performing Cross Validation

```python
neighbors = []
cv_scores = []

from sklearn.model_selection import cross_val_score
for k in range(1,51,2):
    neighbors.append(k)
    knn = KNeighborsClassifier(n_neighbors= k)
    scores = cross_val_score(knn, X_train, y_train , cv = 10, scoring = 'accuracy')
    cv_scores.append(scores.mean())
```

#### Misclassification error versus k

```python
MSE = [1 - x for x in cv_scores]

optimal_k = neighbors[MSE.index(min(MSE))]
print('The optimal number of neighbors is % d ' % optimal_k)
```

```python
# plot misclassification error versus k
plt.figure(figsize = (10, 6))
plt.plot(neighbors, MSE)
plt.xlabel('Number of neighbors')
plt.ylabel('Misclassification Error')
plt.show()
```
