```python
# Importing the Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

```python
# Loading the Data
data = pd.read_csv('Language Detection.csv')
```

```python
# Analyse the Top 5 rows of the Data
data.head()
```

```python
# Let’s have a look at whether this dataset contains any null values
data.isnull().sum()
```

```python
# Now let’s have a look at all the languages present in this dataset
data['language'].value_counts()
```

##### This dataset contains 22 languages with 1000 sentences from each language. This is a very balanced dataset with no missing values, so we can say this dataset is completely ready to be used to train a machine learning model.

#### Language Detection Model


```python
x = np.array(data["Text"])
y = np.array(data["language"])

cv = CountVectorizer()
X = cv.fit_transform(x)
X_train, X_test, y_train, y_test = train_test_split(X, y, 
                                                    test_size=0.33, 
                                                    random_state=42)
```

##### As this is a problem of multiclass classification, we will be using the Multinomial Naïve Bayes algorithm to train the language detection model as this algorithm always performs very well on the problems based on multiclass classification

```python
model = MultinomialNB()
model.fit(X_train,y_train)
model.score(X_test,y_test)
```

```python
# Now let’s use this model to detect the language of a text by taking a user input
user = input("Enter a Text: ")
data = cv.transform([user]).toarray()
output = model.predict(data)
print(output)
```

##### So as you can see that the model performs well. One thing to note here is that this model can only detect the languages mentioned in the dataset.
