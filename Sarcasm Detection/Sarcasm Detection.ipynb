```python
# Importing the Libraries
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB
```

```python
# Loading the Data
data = pd.read_json('Sarcasm Detection.json', lines= True)
```

```python
# Analyse the Top 5 rows of the Data
data.head()
```

##### The “is_sarcastic” column in this dataset contains the labels that we have to predict for the task of sarcasm detection. It contains binary values as 1 and 0, where 1 means sarcastic and 0 means not sarcastic.

```python
# So for simplicity, I will transform the values of this column as “sarcastic” and “not sarcastic” instead of 1 and 0
data["is_sarcastic"] = data["is_sarcastic"].map({0: "Not Sarcasm", 1: "Sarcasm"})
print(data.head())
```

```python
# Now let’s select these columns and split the data into 20% test set and 80% training set

data = data[["headline", "is_sarcastic"]]
x = np.array(data["headline"])
y = np.array(data["is_sarcastic"])

cv = CountVectorizer()
X = cv.fit_transform(x) # Fit the Data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
# Now let's use the Bernoulli Naive Bayes algorithm to train a model for the task of sarcasm detection
model = BernoulliNB()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

```python
# Now let’s use a sarcastic text as input to test whether our machine learning model detects sarcasm or not
user = input("Enter a Text: ")
data = cv.transform([user]).toarray()
output = model.predict(data)
print(output)
```
