```python
# Importing the Libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import BernoulliNB
```

```python
# Loading the Dataset
data = pd.read_csv('Spam Comments Detection.csv')
```

```python
# Analyse the Top 5 rows of the Data
data.head()
```

```python
# We only need the content and class column from the dataset for the rest of the task. So let’s select both the columns
data = data[['CONTENT', 'CLASS']]
```

##### The class column contains values 0 and 1. 0 indicates not spam, and 1 indicates spam.

```python
#  So to make it look better, I will use spam and not spam labels instead of 1 and 0\
data['CLASS'] = data['CLASS'].map({0 : 'Not Spam', 1: 'Spam'})
```

```python
data.head()
```

#### Training a Classification Model


##### As this problem is a problem of binary classification, We will use the Bernoulli Naive Bayes algorithm to train the model

```python
x = np.array(data["CONTENT"])
y = np.array(data["CLASS"])
```

```python
cv = CountVectorizer()
x = cv.fit_transform(x)
xtrain, xtest, ytrain, ytest = train_test_split(x, y, 
                                                test_size=0.2, 
                                                random_state=42)
```

```python
model = BernoulliNB()
model.fit(xtrain, ytrain)
```

```python
model.score(xtest, ytest)
```

```python
# Now let’s test the model by giving spam and not spam comments as input
sample = 'https://github.com/kamran-ansari56?tab=overview&from=2025-04-01&to=2025-04-25'
data = cv.transform([sample]).toarray()
print(model.predict(data))
```

```python
sample = "Lack of information!" 
data = cv.transform([sample]).toarray()
print(model.predict(data)) 
```

#### Summary

Spam comments detection means classifying comments as spam or not spam. Spam comments on social media platforms are the type of comments posted to redirect the user to another social media account, website or any piece of content.
