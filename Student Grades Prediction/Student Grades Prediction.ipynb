```python
# Importing the Libraries
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.utils import shuffle
```

```python
# Loading the Dataset
data = pd.read_csv('Student Grades Prediction.csv')
```

```python
# Analyse the Top 5 rows of the Data
data.head()
```

##### In this dataset the G1 represents the grades of the first period, G2 represents the grades of the second period, and G3 represents the final grades.

```python
# Now let’s prepare the data and let’s see how we can predict the final grades of the students
data = data[["G1", "G2", "G3", "studytime", "failures", "absences"]]
predict = "G3"
x = np.array(data.drop([predict], axis = 1))
y = np.array(data[predict])
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.2)
```

```python
# Now let’s see how to train a linear regression model for the task of student grades prediction
linear_regression = LinearRegression()
linear_regression.fit(xtrain, ytrain)
accuracy = linear_regression.score(xtest, ytest)
print(accuracy)
```

##### The linear regression model gave an accuracy of about 74% which is not bad in this task.

```python
# Now let’s have a look at the predictions made by the students’ grade prediction model
predictions = linear_regression.predict(xtest)
for i in range(len(predictions)):
    print(predictions[x], xtest[x], [ytest[x]])
```
